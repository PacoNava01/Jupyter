###  Generación de Dominios: `np.arange` vs `np.linspace`

Para la creación de lo que conocemos como **"dominios"** de una función, la librería de NumPy ofrece herramientas prácticas que evitan la creación de listas "a mano". Basamos este proceso en dos funciones clave:

---

### 1. `np.arange(start, stop, step)`

* **Parámetros:** 
    * `start`: Inicio de la secuencia.
    * `stop`: Final de la secuencia (valor **no incluido** en el resultado).
    * `step`: El "salto" o incremento entre cada elemento.
* **Uso ideal:** Cuando conoces la distancia exacta que debe haber entre los datos.
* **Ejemplo:** `np.arange(0, 10, 2)` genera `[0, 2, 4, 6, 8]`.



### 2. `np.linspace(start, stop, num)`

* **Parámetros:**
    * `start`: Inicio de la secuencia.
    * `stop`: Final de la secuencia (valor **sí incluido** por defecto).
    * `num`: Número total de elementos que tendrá la lista.
* **Uso ideal:** Cuando necesitas un número específico de muestras para graficar una función suave o recolectar datos de un sensor.
* **Ejemplo:** `np.linspace(0, 10, 5)` genera `[0, 2.5, 5, 7.5, 10]`.

---

###  Comparación Rápida

| Función | Controlas el... | Resultado |
| :--- | :--- | :--- |
| **`np.arange`** | **Espaciado** (`step`) | La cantidad de elementos puede variar según el paso. |
| **`np.linspace`** | **N° de elementos** (`num`) | El espaciado se calcula automáticamente para encajar. |

**Diferencia clave:** En `arange`, el límite superior es **excluyente** (abierto), mientras que en `linspace` es **incluyente** (cerrado). Esto es vital para evitar errores de desfase en tus gráficas.

````python
import numpy as np

# 1. Definición de dominios
# Generamos números de 0 a 10 con un paso de 2 (El 10 queda fuera)
dominio_paso = np.arange(0, 10, 2)

# Generamos exactamente 5 puntos entre 0 y 10 (El 10 queda incluido)
dominio_puntos = np.linspace(0, 10, 5)

# 2. Visualización de resultados con formato limpio
print(f"Resultado np.arange   (paso 2):   {dominio_paso}")
print(f"Resultado np.linspace (5 puntos): {dominio_puntos}")

# 3. Verificación de dimensiones
print("-" * 30) # Línea separadora
son_iguales = len(dominio_paso) == len(dominio_puntos)
print(f"¿Ambos arreglos tienen el mismo tamaño?: {son_iguales}")
print(f"Tamaño de los arreglos: {len(dominio_paso)} elementos")
````

## 🗄️ Creación de Matrices (Arreglos Bidimensionales)

En computación científica, una **matriz** es una estructura de datos organizada en filas ($m$) y columnas ($n$). En NumPy, estas se manejan como objetos `ndarray` de dos dimensiones.

### 1. Creación Manual con `np.array()`
Podemos crear una matriz pasando una lista de listas, donde cada lista interna representa una fila.

* **Sintaxis:** `np.array([[fila1], [fila2]])`
* **Ejemplo:** Una matriz de $2 \times 3$ (2 filas, 3 columnas):
  ```python
  matriz = np.array([[1, 2, 3], [4, 5, 6]])

````python
matriz = np.array([[1,2,3],[4,5,6]])
print(matriz)
````

De igual manera, numpy ofrece funciones para crear matrices comunes sin escribir cada elemento:
* `np.zeros((m, n))`: Crea una matriz llena de ceros. Útil para inicializar variables de almacenamiento.
* `np.ones((m, n))`: Crea una matriz llena de unos.
* `np.eye(n)`: Crea una Matriz Identidad de tamaño $n \times n$ (unos en la diagonal principal y ceros en el resto).
* `np.random.rand(m, n)`: Crea una matriz con valores aleatorios entre 0 y 1.


> Nota técnica: A diferencia de las listas de Python, **todos los elementos de una matriz de `NumPy`** deben ser del mismo tipo (generalmente números enteros o flotantes) para garantizar la velocidad de cálculo.



## Operaciones con matrices
Mediante numpy podemos llevar acabo operaciones algebraicas entre matrices con funciones ya predefinidas y los modulos `np.linalg`

````python
#Definimos unas matrices para trabajar
matriz_A = np.array([[1,2],[3,4]])
matriz_B = np.array([[5,6],[7,8]])


print("Suma y resta de matrices")
#Suma de matrices y resta de matrices
matriz_C = matriz_A + matriz_B #Al ser objetos de numpy ya sabran como sumarse entre estos dos
print("Matriz_C (Suma): \n",matriz_C)
matriz_C = matriz_A - matriz_B 
print("\nMatriz_C (Resta): \n",matriz_C)
print("--"*30)
print("Multiplicación de matrices")

#Multiplicación de matrices
matriz_C = np.dot(matriz_A,matriz_B) #La función dot() ayuda a realizar esto
print("Matriz_C: \n",matriz_C)
print("--"*30)

#Determinante de matrices
print("Determinante de matrices\n")
matriz_C = np.linalg.det(matriz_A)
print("Matriz_C (Basado en la matriz A): \n",matriz_C)
print("--"*30)

#Invertir una matriz
print("Invertir matrices\n")
matriz_C = np.linalg.inv(matriz_A)
print("Matriz_C (Basado en la matriz A): \n",matriz_C)
print("--"*30)

#Transponer matrices
print("Transponer matrices\n")
#Resulta sencillo hacer o conocer matrices transpuestas mediente al tributo de numpy .T
print("Matriz_C (Basado en la matriz A): \n",matriz_A.T)
````

Más allá de la creación de datos, **NumPy** dispone de herramientas de álgebra lineal para resolver sistemas de ecuaciones complejos. Gracias a funciones como `np.linalg.solve()`. Esta capacidad es clave para calcular, por ejemplo, la cinemática de un robot o equilibrar lecturas de múltiples sensores simultáneamente."

# Ejemplo
Para resolver un sistema de ecuaciones lineales mediante `NumPy`, primero debemos representar el problema en su forma matricial $Ax = B$.

**Sistema de ecuaciones:**
$$
\begin{cases} 
x + 2y = 8 \\ 
3x + 4y = 18 
\end{cases}
$$

**Representación matricial:**
Donde $A$ es la matriz de coeficientes y $B$ es el vector de términos independientes:
$$
\begin{bmatrix} 
1 & 2 \\ 
3 & 4 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y 
\end{bmatrix} 
= 
\begin{bmatrix} 
8 \\ 
18 
\end{bmatrix}
$$

Utilizando la función `np.linalg.solve(A, B)`, NumPy aplica métodos numéricos para encontrar los valores de las incógnitas.

````python
# Definimos nuevamente la matriz de coeficientes A
# y, ahora el vector de resultados (B)
A = np.array([[1, 2], [3, 4]])
B = np.array([8, 18])

# Resolvemos el sistema: Ax = B
solucion = np.linalg.solve(A, B)

print(f"La solución es: x = {solucion[0]}, y = {solucion[1]}")
# Resultado esperado: x=2, y=3
````

````python
import matplotlib.pyplot as plt #La forma más sencilla de hacerlo
x = np.linspace(0,25,100)

#Definimos nuestra función
y = np.sin(x)
y1 = np.cos(x)


#Esta linea siempre de ir primero para modificar el tamaño del grafico
plt.figure(figsize=(10,2))

#Funciones (lw = line weight| ms = marker size(tamaño bolitas))
plt.plot(x,y,'o-',label = "sin(x)",color = "darkgreen", lw = 1)
plt.plot(x,y1,'--',label= "cos(x)",color= "red", lw = 1)

#posicion y ajustes de la label 
plt.legend(loc= 'upper right',fontsize = 8, edgecolor = 'steelblue')
#Titulos y Nombres ejes
plt.title('Grafica Sencilla',fontsize = 12)
plt.xlabel("tiempo",fontsize=8)
plt.ylabel("Rango",fontsize=8)

#Limites de los ejes
plt.xlim(0,25)
plt.ylim(-1.5,1.5)



#Tamaño de los números de los ejes
plt.tick_params(axis = 'both', labelsize = 9.5)

#mostrar
plt.grid()
plt.show()
````

# Lo mismo pero separado (subplots)
Imaginando que las graficas nos las piden por separado, que quieren que en una sola figura esten el seno y el coseno pero en graficas aparte 
* subplots(cantidad de graficos en x, cantidad de graficos en y, número del grafico actual )

- Un ax/axe es un cuadrito individual
- La figura (fig) es la imagen en conjunto, son todos los axes

````python
#Datos
x = np.linspace(0,25,50)
y = np.sin(x)
y2 = np.cos(x)
y3 = x
y4 = x**(1/2)

#Se definen cuantos graficos seran y de qué tamaño
fig,axes = plt.subplots(2,2,figsize=(10,6))
#2,2: Indica la cantidad de graficas que tendremos en número de filas y columnas
#fig: Es el total de la imagen
#axes: Son las graficas que iran dentro de esa gran imagen
#Ahora se programa cada grafico, parecido a lo realizado anteriormente, pero  individual


#Primero
ax = axes[0][0]
ax.plot(x,y,'--',color = 'navy',lw = 1)
ax.set_xlabel('Tiempo (s)',fontsize=8)
ax.set_ylabel('Sen(x)',fontsize=8)
ax.tick_params(axis = 'both',labelsize = 8)
ax.grid()

#Segundo
ax = axes[0][1]
ax.plot(x,y2-y4,'-o',color="green",lw = 0.5,ms = 1)
ax.set_ylabel('Cos(x)', fontsize = 8)
ax.set_xlabel('Tiempo (s)',fontsize=8)
ax.grid()


#Tercero
ax = axes[1][0]
ax.plot(x,y3,'3',color= "#BF00BF", lw = 0.5,ms = 2)
ax.set_ylabel('$y = x$', fontsize = 8)
ax.set_xlabel('Tiempo (s)',fontsize=8)
ax.grid()


#Cuarto
ax = axes[1][1]
ax.plot(x,y4)
ax.set_ylabel('$y = x^{1/2}$', fontsize = 8)
ax.set_xlabel('Tiempo (s)',fontsize=8)
ax.grid()




#mostrar
#Se ajustan los parametros de los subplots para que no esten amontonados, los cambiamos con 
plt.subplots_adjust(hspace=0.4,wspace=0.45)
plt.show()
````

# Datos experimentales
<font color = red >`errorbar`</font> es una herramienta que nos ayuda a graficar contemplando ya las barras de error, algunos de los argumentos son:

* fmt - Tipo de linea o marca
* capthick - Indica el ancho de las lineas en los extremos (caps)
* capsize - Tamaño del cap
* elinewidth - Grosor de la linea de error

````python
rng = np.random.default_rng()
#Datos Muestra 1
x1 = rng.uniform(0,10,size = 6)
x1 = np.round(x1,2) #Se redondea a 2 decimales
y1 = rng.uniform(0,10,size = 6)
y1 = np.round(x1,2)
dest1 = rng.uniform(0,9.9,size = 6) #desviación estandar

#Datos Muestra 2
x2 = rng.uniform(0,10,size = 6)
x2 =  np.round(x2,2) #Se redondea a 2 decimales
y2 = rng.uniform(0,10,size = 6)
y2 = np.round(x2,2)
dest2 = rng.uniform(0,9.9,size = 6) #desviación estandar

#Datos Muestra 3
x3 =  rng.uniform(0,10,size = 6)
x3 = np.round(x3,2) #Se redondea a 2 decimales
y3 = rng.uniform(0,10,size = 6)
y3 = np.round(x3,2)
dest3 = rng.uniform(0,9.9,size = 6) #desviación estandar


#Se crea una figura y un solo subplot
fig,axes = plt.subplots(figsize = (10,5))

#Grafica de datos
axes.errorbar(x1,y1,dest1,fmt = 'o--',capsize=6,elinewidth=0.75,
              capthick = 0.75,color = 'red', label = "Muestra 1")
axes.errorbar(x2,y2,dest2,fmt = 'o--',capsize=6,elinewidth=0.75,
              capthick = 0.75,color = 'deeppink', label = "Muestra 2")
axes.errorbar(x3,y3,dest3,fmt = 'o--',capsize=6,elinewidth=0.75,
              capthick = 0.75,color = 'steelblue', label = "Muestra 3")
plt.legend(loc = 'upper left',fontsize = 12,edgecolor = 'darkgray')

axes.set_xlabel('Tiempo (s)',fontsize=15)
axes.set_ylabel('Distancia',fontsize=15)
axes.tick_params(axis = 'both',labelsize = 10)


#Titulos y Nombres ejes
plt.grid(True)
plt.show()
````

# Graficos de barras

````python
#Datos 
categorias = ['A', 'B' , 'C','D','E']
valores = [5,7,3,8,4]

colores = ['steelblue','slategrey','teal','slateblue']

#Total y porcentajes
Total = sum(valores)
porcentajes = [valor/Total * 100 for valor in valores]


#Grafico
plt.figure(figsize=(10,5))
plt.bar(categorias,valores,color = colores)

#Ajustar ejes
plt.xlabel('Categorias',fontsize = 10)
plt.ylabel('valores',fontsize = 10)
plt.tick_params(axis = 'both',labelsize = 8.5)
plt.ylim(0,9)

#Valores de barras
for i, (valor,porcentaje) in enumerate(zip(valores,porcentajes)):
    plt.text(i,valor + 0.1, f'{valor}\n{porcentaje:.1f}',
    ha = 'center', fontsize = 8)

#Mostrar
plt.show()
````
