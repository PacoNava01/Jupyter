## Nivel 1: El Calentamiento (Sistemas de $2 \times 2$)

En robótica 2D (como un robot movil en el piso), muchas veces trabajarás con planos $XY$.El Problema:Imagina que quieres encontrar el punto de intersección de dos trayectorias dadas por:

$$
\begin{cases} 
2x + y = 10 \\ 
x - 3y = -2
\end{cases}
$$

----
## **Objetivo**
* Representa esto en la forma $Ax = B$.
* Usa np.linalg.solve() para hallar $x$ e $y$.


$$
\begin{bmatrix} 
2 & 1 \\ 
1 & -3 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y 
\end{bmatrix} 
= 
\begin{bmatrix} 
10 \\ 
-2 
\end{bmatrix}
$$

````python
#Importamos la libreria necesaria para realizar los ejercicios
import numpy as np
````

````python
#Comenzamos definiendo las matrices
A = np.array([[2,1],[1,-3]])
B = np.array([10,-2])

sol = np.linalg.solve(A,B)
print(f"El vector solución es:  x ={sol[0]} y ={sol[1]}")
````

## Nivel 3: El Robot en el Espacio 3D (Matrices de Transformación)
Los robots no solo se mueven, también rotan. Aunque no es un "sistema de ecuaciones" clásico de hallar $x$, la resolución de matrices es clave para invertir movimientos.

---

**El Problema**:
Tienes una matriz de rotación $R$ (que gira un vector 45° en el eje Z) y un vector de posición $P$

````python
theta = np.radians(45) #Angulo de rotación
R = np.array([[np.cos(theta), -np.sin(theta), 0],
              [np.sin(theta),  np.cos(theta), 0],
              [0,              0,             1]]) #Matriz de rotación clasica
B = np.array([1.5, 2.0, 0]) # Posición final deseada
````

**Objetivo**

 * Si $R \cdot x = B$, ¿cuál era la posición original $x$ del sensor antes de rotar?
 
 * Usa NumPy para resolver este sistema.

````python
#Aplicamos la matriz de rotación inversa
R_inv = np.linalg.inv(R)
x = np.dot(R_inv,B)
print(x)

# Tambien podemos aplicar dado que buscando x, estamos practicamente buscando soluciónar un S.EC.D
x = np.linalg.solve(R,B)
x
````

## Nivel 4

A veces, en robótica, tienes más ecuaciones que incógnitas (por ejemplo, 3 sensores GPS para determinar una posición $X, Y$). Esto se resuelve con **Mínimos Cuadrados**.

---

El Problema:Tienes estas lecturas que "chocan" un poco entre sí:
- $x + y = 2$
- $x - y = 0$
- $2x + y = 3.1$ (Hay un pequeño error de lectura aquí)
# Objetivo

* Si usas `np.linalg.solve()`, NumPy lanzará un error porque la matriz no es cuadrada.

* Investiga y utiliza `np.linalg.lstsq()` para encontrar la "mejor solución posible".

### Primero la convertimos en algo de la forma $Ax = B$

$$
\begin{bmatrix} 
1 & 1 \\ 
1 & -1 \\
2 & 1 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y
\end{bmatrix} 
= 
\begin{bmatrix} 
1 \\ 
-2 \\
0 
\end{bmatrix}
$$

````python
#Creamos las matrices
A  = np.array([[1,1],[1,-1],[2,1]])
B = np.array([[1],[-2],[0]])

sol = np.linalg.solve(A,B)

````

````python
#Creamos las matrices
A  = np.array([[1,1],[1,-1],[2,1]])
B = np.array([[1],[-2],[0]])

#La manera de solucionar sistemas sobredeterminados, como es este caso es mediante np.linalg.lstq() (least-squares solution)
sol = np.linalg.lstsq(A,B,rcond= None)
sol
````

Como se observa, la solución nos arroja una lista de arreglos, por lo cual es donde entran los outputs de nuestra solución
* #  Solución
`solution` indica los valores que mejor se ajustan a las incógnitas.
* #  Residuos
`residuals` Mide la distancia entre los valores predichos y los reales. Un residuo menor significa que la solución se ajusta mejor.
* #  Rango
Indican la estabilidad de la solución. Si el rango es menor que el número de variables, es posible que el sistema no tenga una solución única.
* #  Valores singulares
Ayudan a determinar si el sistema es estable y resoluble. Aunque no se utilicen directamente, pueden indicar estabilidad numérica.

````python
#Con esta información obtenida tendremos que la respuesta la deberemos de pedir de la forma
sol,res,rango,val_sing = np.linalg.lstsq(A,B,rcond= None)
sol
````

## Nivel 5

Para un robot diferencial de 4 ruedas, la relación entre la velocidad de las ruedas ($v_l, v_r$) y la velocidad lineal ($v$) y angular ($\omega$) del robot es:

$$\begin{cases} 
v = \frac{r}{2}(v_r + v_l) \\
\omega = \frac{r}{L}(v_r - v_l) 
\end{cases}$$

Donde:

$r$ = radio de la rueda (ej. 0.05m).

$L$ = distancia entre ruedas (ej. 0.3m).

---
## Objetivo:
Si quieres que tu robot se mueva a $v = 1.0 \, m/s$ y rote a $\omega = 0.5 \, rad/s$: 

* Plantea el sistema de ecuaciones donde las incógnitas son $v_r$ y $v_l$.
* Resuélvelo con `NumPy` para saber a qué velocidad setear tus motores.

Sustituimos los valores de r, L , v y $\omega$ obteniendo el sistema de ecuaciones

$$\begin{cases} 
1.0 = \frac{0.005}{2}v_r + \frac{0.005}{2}v_l \\
0.5 = \frac{0.005}{0.03}v_r -\frac{0.005}{0.3L} v_l 
\end{cases}$$





$$
\begin{bmatrix} 
1 \\ 
0.5
\end{bmatrix}
 = \begin{bmatrix} 
\frac{0.005}{2} & \frac{0.005}{2} \\ 
\frac{0.005}{0.03} & -\frac{0.005}{0.03}
\end{bmatrix} 
\begin{bmatrix} 
v_r \\ 
v_l
\end{bmatrix} 
$$ 


````python
#Datos
r = 0.005 #m
L = 0.3 #m

x = r/2
y = r/L

B = np.array([[1.0],[0.5]])
A = np.array([[x,x],[y,-y]])

sol = np.linalg.solve(A,B)
sol
````

Con esta información ahora sabemos que podemos conseguir este resultado de velocidad lineal y velocidad angular, llevando a las llantas derecha e izquierda a las velocidades $v_r = 215 m/s$ y $v_l = 185m/s$ respectivamente
