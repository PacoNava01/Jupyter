## Nivel 1: El Calentamiento (Sistemas de $2 \times 2$)

En robótica 2D (como un robot movil en el piso), muchas veces trabajarás con planos $XY$.El Problema:Imagina que quieres encontrar el punto de intersección de dos trayectorias dadas por:

$$
\begin{cases} 
2x + y = 10 \\ 
x - 3y = -2
\end{cases}
$$

----
## **Objetivo**
* Representa esto en la forma $Ax = B$.
* Usa np.linalg.solve() para hallar $x$ e $y$.


$$
\begin{bmatrix} 
2 & 1 \\ 
1 & -3 
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y 
\end{bmatrix} 
= 
\begin{bmatrix} 
10 \\ 
-2 
\end{bmatrix}
$$

```python
#Importamos la libreria necesaria para realizar los ejercicios
import numpy as np
```

```python
#Comenzamos definiendo las matrices
A = np.array([[2,1],[1,-3]])
B = np.array([10,-2])

sol = np.linalg.solve(A,B)
print(f"El vector solución es:  x ={sol[0]} y ={sol[1]}")
```

# Nivel 2: Control de Motores (Sistemas de $3 \times 3$)

Supongamos que tienes un sistema de control de energía para tres motores. La potencia total y las relaciones de voltaje te dan estas ecuaciones:
$$\begin{cases} 
3x + 2y - z = 1 \\
2x - 2y + 4z = -2 \\
-x + 0.5y - z = 0 
\end{cases}$$

---
## **Objetivo**
* Resuelve el sistema.
* Extra: Verifica el resultado multiplicando la matriz $A$ por tu solución $x$ usando el operador `@` (`A @ x_sol`). Debería darte exactamente $B$.

$$
\begin{bmatrix} 
3 & 2 & -1 \\ 
2 & -2 & +4\\
-1 & 0.5 & -1
\end{bmatrix} 
\begin{bmatrix} 
x \\ 
y \\
z
\end{bmatrix} 
= 
\begin{bmatrix} 
1 \\ 
-2 \\
0 
\end{bmatrix}
$$

```python
A = np.array([[3,2,-1],[2,-2,4],[-1,0.5,-1]])
B = np.array([1,-2,0])

sol = np.linalg.solve(A,B)

#Extra = np.dot(A,sol)
Extra = A @ sol #Manera alternativa para realizar producto punto entre dos vectores/matrices
Extra
```

## Nivel 3: El Robot en el Espacio 3D (Matrices de Transformación)
Los robots no solo se mueven, también rotan. Aunque no es un "sistema de ecuaciones" clásico de hallar $x$, la resolución de matrices es clave para invertir movimientos.

---

**El Problema**:
Tienes una matriz de rotación $R$ (que gira un vector 45° en el eje Z) y un vector de posición $P$

```python
theta = np.radians(45) #Angulo de rotación
R = np.array([[np.cos(theta), -np.sin(theta), 0],
              [np.sin(theta),  np.cos(theta), 0],
              [0,              0,             1]]) #Matriz de rotación clasica
B = np.array([1.5, 2.0, 0]) # Posición final deseada
```

**Objetivo**

 * Si $R \cdot x = B$, ¿cuál era la posición original $x$ del sensor antes de rotar?
 
 * Usa NumPy para resolver este sistema.

```python
#Aplicamos la matriz de rotación inversa
R_inv = np.linalg.inv(R)
x = np.dot(R_inv,B)
print(x)

# Tambien podemos aplicar dado que buscando x, estamos practicamente buscando soluciónar un S.EC.D
x = np.linalg.solve(R,B)
x
```
